```python
max_fuel_eff_asia = df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
print(max_fuel_eff_asia)
```

```python
median_hp = df['horsepower'].median()
print(median_hp)
```

```python
most_freq_hp = df['horsepower'].mode()[0]
print(most_freq_hp)
```

```python
df['horsepower'] = df['horsepower'].fillna(most_freq_hp)
print(df['horsepower'].isnull().sum())
```

```python
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].iloc[:7].values
```

```python
XTX = X.T @ X
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
```

```python
result = w.sum()
print(result)
```
